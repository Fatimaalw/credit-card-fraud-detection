```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("creditcard.csv")

print("\n=== DATA UNDERSTANDING ===")
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())

print("\nSummary Statistics:")
display(df.describe())

print("\n=== DATA TYPES ===")
print(df.dtypes)

display(df.head())

print("=== BEFORE CLEANING ===")
print("Shape:", df.shape)
print("Missing Values:", df.isnull().sum().sum())
print("Duplicates:", df.duplicated().sum())

print("\n=== INVALID VALUES ===")
print("Negative Amount:", (df["Amount"] < 0).sum())
print("Negative Time:", (df["Time"] < 0).sum())
print("Invalid Class:", (~df["Class"].isin([0, 1])).sum())

df = df.dropna()
df = df.drop_duplicates()
df = df[(df["Amount"] >= 0) & (df["Time"] >= 0)]
df = df[df["Class"].isin([0, 1])]


print("\n=== AFTER CLEANING ===")
print("Shape:", df.shape)
print("Missing Values:", df.isnull().sum().sum())
print("Duplicates:", df.duplicated().sum())

Q1 = df["Amount"].quantile(0.25)
Q3 = df["Amount"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df["Amount"] < lower) | (df["Amount"] > upper)]

print("\n=== OUTLIERS ===")
print("Amount Outliers:", len(outliers))
print("Fraud Among Outliers:", (outliers["Class"] == 1).sum())



plt.boxplot(df["Amount"])
plt.title("Transaction Amount Outliers")
plt.ylabel("Amount")
plt.show()

df.to_csv("creditcard_clean.csv", index=False)
print("Cleaned dataset saved!")
```

## Data Understanding
The dataset contains 284,807 transactions and 31 columns.
The target variable is Class (0 = Normal, 1 = Fraud).

## Data Cleaning
The dataset had no missing or invalid values.
We removed 1,081 exact duplicate rows.
After cleaning, 283,726 transactions remained.

## Outlier Detection
We used the IQR method to detect unusual transaction amounts.
We found 31,685 Amount outliers, including 87 fraudulent transactions.
We kept the outliers because they may contain useful fraud patterns.

## Data Limitations
The dataset is highly imbalanced, with only 473 fraudulent transactions.
The original meanings of V1–V28 are unknown.
Removing duplicates also removed 19 fraud records.

## Feature Engineering

Two new features were created from the Amount and Time columns to explore additional patterns in fraudulent transactions.

```python
import numpy as np

# Create new features
df["Log_Amount"] = np.log1p(df["Amount"])
df["Hour_Offset"] = ((df["Time"] // 3600) % 24).astype(int)

# Display new features
display(df[["Amount", "Log_Amount", "Time", "Hour_Offset"]].head())

# Check dataset shape
print("Dataset Shape:", df.shape)

# Save dataset
df.to_csv("creditcard_features.csv", index=False)

print("Feature engineered dataset saved!")
```

### Feature Engineering Findings

Two new features were created: Log_Amount and Hour_Offset. Log_Amount applies a logarithmic transformation to transaction amounts to reduce the influence of very large values. Hour_Offset represents the transaction's relative position within a 24-hour cycle. These features may help detect fraud, but their usefulness will be evaluated during modelling.

## Stage 5 – Model Selection and Development

**Goal:** build and compare several models for fraud detection (target `Class`), starting from a baseline, using a leakage-free workflow.

**Design decisions**
- **Features:** V1–V28, `Log_Amount`, `Hour_Offset`. Raw `Amount` and `Time` are dropped because `Log_Amount` and `Hour_Offset` replace them.
- **Split:** 80/20 stratified train/test split, done *before* any scaling or resampling. The test set is only used once, for the final comparison.
- **Imbalance:** handled with `class_weight` inside the models (no oversampling), so no synthetic rows can leak into validation or test data.
- **Scaling:** `StandardScaler` is inside a Pipeline for Logistic Regression, so it is fitted on training folds only.
- **Validation:** 5-fold stratified cross-validation on the training set only.
- **Metrics:** PR-AUC (main metric), ROC-AUC, Precision, Recall, F1. Accuracy is not used because 99.83% of the data is normal.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_validate
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, HistGradientBoostingClassifier
from sklearn.metrics import (average_precision_score, roc_auc_score, precision_score,
                             recall_score, f1_score, confusion_matrix, precision_recall_curve)

RANDOM_STATE = 42

df = pd.read_csv("creditcard_features.csv")
feature_cols = [f"V{i}" for i in range(1, 29)] + ["Log_Amount", "Hour_Offset"]
X = df[feature_cols]
y = df["Class"]
print("X shape:", X.shape, "| fraud rate: %.4f%%" % (y.mean() * 100))
```

```python
# Stratified split BEFORE any preprocessing (prevents data leakage)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, stratify=y, random_state=RANDOM_STATE
)
print("Train:", X_train.shape, "| fraud:", y_train.sum(), "(%.3f%%)" % (y_train.mean() * 100))
print("Test :", X_test.shape,  "| fraud:", y_test.sum(),  "(%.3f%%)" % (y_test.mean() * 100))
```

### Candidate models
| Model | Why it is included |
|---|---|
| Dummy (most frequent) | Baseline: shows what a model with no skill scores. |
| Logistic Regression | Simple, fast, interpretable linear model. |
| Random Forest | Non-linear ensemble of trees; robust to outliers and does not need scaling. |
| Hist Gradient Boosting | Strong boosting model for tabular data; usually the best performer on this kind of dataset. |

```python
models = {
    "Baseline (Dummy)": DummyClassifier(strategy="most_frequent"),
    "Logistic Regression": Pipeline([
        ("scaler", StandardScaler()),
        ("clf", LogisticRegression(class_weight="balanced", max_iter=1000, random_state=RANDOM_STATE)),
    ]),
    "Random Forest": RandomForestClassifier(
        n_estimators=100, max_depth=12, class_weight="balanced_subsample",
        n_jobs=-1, random_state=RANDOM_STATE),
    "Hist Gradient Boosting": HistGradientBoostingClassifier(
        class_weight="balanced", random_state=RANDOM_STATE),
}
```

### Cross-validation (training set only)
Used to compare the models fairly and choose the final model without touching the test set. This cell can take a few minutes.

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
scoring = {
    "pr_auc": "average_precision",
    "roc_auc": "roc_auc",
    "precision": "precision",
    "recall": "recall",
    "f1": "f1",
}

cv_rows = []
for name, model in models.items():
    res = cross_validate(model, X_train, y_train, cv=cv, scoring=scoring, n_jobs=1)
    cv_rows.append({
        "Model": name,
        "PR-AUC": res["test_pr_auc"].mean(),
        "ROC-AUC": res["test_roc_auc"].mean(),
        "Precision": res["test_precision"].mean(),
        "Recall": res["test_recall"].mean(),
        "F1": res["test_f1"].mean(),
        "PR-AUC std": res["test_pr_auc"].std(),
    })
    print("done:", name)

cv_results = pd.DataFrame(cv_rows).set_index("Model").round(4)
display(cv_results)
```

### Test-set evaluation
Each model is refitted on the full training set and evaluated once on the untouched test set.

```python
test_rows, probas, cms = [], {}, {}
for name, model in models.items():
    model.fit(X_train, y_train)
    pred = model.predict(X_test)
    # DummyClassifier has predict_proba too (constant), so this works for all models
    proba = model.predict_proba(X_test)[:, 1]
    probas[name] = proba
    cms[name] = confusion_matrix(y_test, pred)
    test_rows.append({
        "Model": name,
        "PR-AUC": average_precision_score(y_test, proba),
        "ROC-AUC": roc_auc_score(y_test, proba),
        "Precision": precision_score(y_test, pred, zero_division=0),
        "Recall": recall_score(y_test, pred, zero_division=0),
        "F1": f1_score(y_test, pred, zero_division=0),
    })

test_results = pd.DataFrame(test_rows).set_index("Model").round(4)
display(test_results)

for name, cm in cms.items():
    tn, fp, fn, tp = cm.ravel()
    print(f"{name:25s} TN={tn:6d}  FP={fp:5d}  FN={fn:4d}  TP={tp:4d}")
```

```python
# Precision-Recall curves
plt.figure(figsize=(8, 6))
for name, proba in probas.items():
    if name.startswith("Baseline"):
        continue
    p, r, _ = precision_recall_curve(y_test, proba)
    plt.plot(r, p, label=f"{name} (PR-AUC = {average_precision_score(y_test, proba):.3f})")
plt.axhline(y_test.mean(), color="grey", linestyle="--", label="Baseline (fraud rate)")
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curves on the Test Set")
plt.legend()
plt.grid(True)
plt.show()

# Save results for the report / GitHub
cv_results.to_csv("stage5_cv_results.csv")
test_results.to_csv("stage5_test_results.csv")
```

```python
# Select the final model by cross-validated PR-AUC (not by test score)
best_name = cv_results.drop(index="Baseline (Dummy)")["PR-AUC"].idxmax()
print("Selected model for Stage 6 tuning:", best_name)
```
